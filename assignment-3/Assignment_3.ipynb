```python
data2 = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(data2)
employees.to_csv('employees.csv', index=False)
df = pd.read_csv("employees.csv")
print(df)
```

```python
# Shape

print("Shape:", df.shape)
```

```python
# Summary of DataFrame

df.info()
```

```python
# Descriptive statistics

print(df.describe())
```

```python
# First 5 rows and last 3 rows

print("First 5 rows:")
print(df.head(5))

print("")

print("\nLast 3 rows:")
print(df.tail(3))
```

```python
# i. Average salary
averageSalary = df["Salary"].mean()

# ii. Total bonus
totalBonus = df["Bonus"].sum()

# iii. Youngest employee's age
youngestAge = df["Age"].min()

# iv. Highest performance rating
highestRating = df["Rating"].max()

print("Average Salary:", averageSalary)
print("Total Bonus:", totalBonus)
print("Youngest Employee's Age:", youngestAge)
print("Highest Performance Rating:", highestRating)
```

```python
# Sort Descending by Salary

df_sorted = df.sort_values(by="Salary", ascending=False)

print(df_sorted)
```

```python
# Add Category Performance

def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

df["Performance_Category"] = df["Rating"].apply(performance_category)

print(df)
```

```python
# Identify Missing values

print(df.isnull().sum())
```

```python
df.rename(columns={"Employee_ID": "ID"}, inplace=True)

print(df)
```

```python
# i. More than 5 years of experience
print("Employees with more than 5 years of experience:")
print(df[df["Years_of_Experience"] > 5])

# ii. Employees belonging to IT department
print("\nEmployees in IT department:")
print(df[df["Department"] == "IT"])
```

```python
# Tax is 10% of salary:

df["Tax"] = df["Salary"] * 0.10

print(df)
```

```python
# Save in a new CSV

df.to_csv("employees_modified.csv", index=False)

print("Modified DataFrame saved as employees_modified.csv")
```
