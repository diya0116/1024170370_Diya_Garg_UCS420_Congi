```python
# Assignment 5.2: WAP to add all numbers divisible by 7 and 9 from 1 to n and n is given by the user.

n = int(input("Enter a number: "))
sum = 0

for i in range(1, n + 1):
    if i % 7 == 0 and i % 9 == 0:
        sum = sum + i

print("Sum is:", sum)
```

```python
# Assignment 5.3: WAP to add all prime numbers from 1 to n and n is given by the user.

n = int(input("Enter a number: "))
sum = 0

for num in range(2, n + 1):
    is_prime = True

    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        sum = sum + num

print("Sum of prime numbers is:", sum)
```

```python
# Assignment 6.1: WAP using function that add all odd numbers from 1 to n, n is given by the user.

def add_odd(n):
    sum = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            sum = sum + i

    return sum


n = int(input("Enter a number: "))
print("Sum of odd numbers is:", add_odd(n))
```

```python
# Assignment 6.2: WAP using function that add all prime numbers from 1 to n, n given by the user.

def addPrime(n):
    sum = 0

    for num in range(2, n + 1):
        isPrime = True

        for i in range(2, num):
            if num % i == 0:
                isPrime = False
                break

        if isPrime:
            sum = sum + num

    return sum


n = int(input("Enter a number: "))
print("Sum of prime numbers is:", addPrime(n))
```
