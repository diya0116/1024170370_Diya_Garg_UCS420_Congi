```python
# 1. Take your roll number. Extract its individual digits, and multiply each digit by 10 to form your list L (10 numbers in list now).
# i. Print L

roll_number = input("Enter your roll number: ")
digits = [int(digit) for digit in roll_number if digit.isdigit()]
L = [digit * 10 for digit in digits]

print(L)
```

```python
# ii. Add two numbers of your choice to L — one using append() and one using insert() at a specific position. Print L after each and explain in a comment what changed.

L.append(50)
print(L)

L.insert(2, 70)
print(L)
```

```python
# iii. Remove two elements from L — one using remove() and one using pop(). Print L after each.

L.remove(50)
print(L)

L.pop()
print(L)
```

```python
# iv. Sort L in ascending order using sort(), then descending order using sort(reverse=True). Print both.

L.sort()
print(L)

L.sort(reverse=True)
print(L)
```

```python
# v. Use slicing to print the first three elements and the last three elements of L in a single line each.

print(L[:3])
print(L[-3:])
```

```python
# vi. Use list comprehension to create a new list containing only the elements of L that are greater than the average of L.

average = sum(L) / len(L)
new_list = [x for x in L if x > average]

print(new_list)
```

```python
# 3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating your numbers.
# i. Generate a list of 100 random numbers between 100 and 900 (inclusive).

import random

roll_number = int(input("Enter your roll number: "))
random.seed(roll_number)

numbers = [random.randint(100, 900) for i in range(100)]

print(numbers)
```

```python
# ii. Count and print all odd numbers in the list.

odd_numbers = [x for x in numbers if x % 2 != 0]

print(odd_numbers)
print(len(odd_numbers))
```

```python
# iii. Count and print all even numbers in the list.

even_numbers = [x for x in numbers if x % 2 == 0]

print(even_numbers)
print(len(even_numbers))
```

```python
# iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension.

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True

prime_numbers = [x for x in numbers if is_prime(x)]

print(prime_numbers)
print(len(prime_numbers))
```

```python
# v. Print the number that occurs most frequently in your list, and how many times it occurs.

most_frequent = max(numbers, key=numbers.count)

print(most_frequent)
print(numbers.count(most_frequent))
```

```python
# 4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for each digit} and set B = {digit × 9 for each digit}. Print both sets.

roll_number = input("Enter your roll number: ")
digits = [int(digit) for digit in roll_number if digit.isdigit()]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print(A)
print(B)
```

```python
# vi. Find the union of A and B — all unique values across both sets.

print(A.union(B))
```

```python
# vii. Find the intersection of A and B — values common to both.

print(A.intersection(B))
```

```python
# viii. Find A − B and B − A separately (not combined) using difference().

print(A.difference(B))
print(B.difference(A))
```

```python
# ix. Find the symmetric difference of A and B.

print(A.symmetric_difference(B))
```

```python
# x. Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset().

print(A.issubset(B))
print(B.issuperset(A))
```

```python
# xi. Ask the user to input a value X. Remove X from set A if present using discard() (not remove()).

X = int(input("Enter a value: "))

A.discard(X)
print(A)
```

```python
# 5. Consider the following dictionary:

my_dict = {
    "name": "Diya Garg",
    "roll_no": "1024170370",
    "branch": "Computer Science",
    "age": 21,
    "city": "Patiala"
}

print(my_dict)
```

```python
# i. Rename the key “city” to “location” without changing its value.

my_dict["location"] = my_dict.pop("city")

print(my_dict)
```

```python
# ii. Add a new key “cgpa” with your CGPA as its value.

my_dict["cgpa"] = 0.0

print(my_dict)
```

```python
# iii. Update the value of “age” by increasing it by 1.

my_dict["age"] += 1

print(my_dict)
```

```python
# iv. Delete the “branch” key using two different methods in two separate copies of the dictionary: once with pop() and once with the del keyword.

dict1 = my_dict.copy()
dict2 = my_dict.copy()

dict1.pop("branch")
print(dict1)

del dict2["branch"]
print(dict2)
```

```python
# v. Iterate over your dictionary using .items() and print each key-value pair as “key → value”.

for key, value in my_dict.items():
    print(key, "→", value)
```

```python
# vi. Before accessing a key called “email” (which doesn't exist in your dictionary), check for its existence using the in keyword and print a safe fallback message instead of letting the program crash.

if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email not found")
```

```python
# vii. Create a second dictionary friend_dict with the same 5 original keys but a friend's (fictional) details. Merge it with your own using the {**dict1, **dict2} syntax.

friend_dict = {
    "name": "Devam",
    "roll_no": "1024170900",
    "branch": "Computer Science",
    "age": 21,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print(merged_dict)
```

```python
# viii. Use dictionary comprehension to create a new dictionary containing only the key-value pairs from your original dictionary where the value is a string.

string_dict = {key: value for key, value in my_dict.items() if isinstance(value, str)}

print(string_dict)
```
