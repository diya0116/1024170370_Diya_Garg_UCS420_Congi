```python
# Question 2 (b-i)

import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

most_frequent = np.bincount(x).argmax()
indices = np.where(x == most_frequent)[0]

print("Array :", x)
print("Most Frequent Value :", most_frequent)
print("Indices :", indices)
```

```python
# Question 2 (b-ii)

import numpy as np

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

most_frequent = np.bincount(y).argmax()
indices = np.where(y == most_frequent)[0]

print("Array :", y)
print("Most Frequent Value :", most_frequent)
print("Indices :", indices)
```

```python
# Question 4

import numpy as np

diya = np.linspace(10, 100, 25)

print("Array :")
print(diya)

print("\nDimensions :", diya.ndim)
print("Shape :", diya.shape)
print("Total Elements :", diya.size)
print("Data Type :", diya.dtype)
print("Total Bytes :", diya.nbytes)

transpose_array = diya.reshape(25, 1)

print("\nTranspose using reshape():")
print(transpose_array)

print("\nUsing T attribute:")
print(diya.T)
```

```python
# Question 5

import numpy as np

ucs420_diya = np.array([[10, 20, 30, 40],
                        [50, 60, 70, 80],
                        [90, 15, 20, 35]])

print("Original Array :")
print(ucs420_diya)

print("\nMean :", np.mean(ucs420_diya))
print("Median :", np.median(ucs420_diya))
print("Maximum :", np.max(ucs420_diya))
print("Minimum :", np.min(ucs420_diya))
print("Unique Elements :", np.unique(ucs420_diya))

reshaped_ucs420_diya = ucs420_diya.reshape(4, 3)

print("\nReshaped Array (4 x 3):")
print(reshaped_ucs420_diya)

resized_ucs420_diya = np.resize(ucs420_diya, (2, 3))

print("\nResized Array (2 x 3):")
print(resized_ucs420_diya)
```
