```python
# Import NumPy

import numpy as np

print("NumPy imported successfully!")
```

```python
# Python List vs NumPy Array

# Python List
py_list = [10, 20, 30, 40]

# NumPy Array
np_array = np.array([10, 20, 30, 40])

print("Python List:", py_list)
print("NumPy Array:", np_array)

print("\nType of List:", type(py_list))
print("Type of NumPy Array:", type(np_array))
```

```python
# 1D, 2D Arrays

# 1D Array
a1 = np.array([1, 2, 3, 4])

# 2D Array
a2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])


print("1D Array:\n", a1)
print("\n2D Array:\n", a2)
```

```python
# Array Attributes

a = np.array([10, 20, 30, 40, 50])

print("Array:", a)

print("dtype :", a.dtype)
print("size  :", a.size)
print("ndim  :", a.ndim)
print("shape :", a.shape)
print("nbytes:", a.nbytes)
```

```python
# Indexing

arr = np.array([20, 1, 2, 3, 4])

print("Original Array:", arr)

print("\nFirst element:", arr[0])
print("Third element:", arr[2])
print("Last element:", arr[-1])
```

```python
# Modifying Elements

arr = np.array([20, 1, 2, 3, 4])

print("Before:", arr)

arr[0] = 100
arr[4] = 0

print("After :", arr)
```

```python
# Slicing

arr = np.array([100, , 0])
1, 2, 3
print("Original:", arr)

print("arr[1:4] :", arr[1:4])
print("arr[:3]  :", arr[:3])
print("arr[2:]  :", arr[2:])
print("Reverse  :", arr[::-1])
```

```python
# 2D Array Indexing

arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(arr)

print("\n1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
print("Entire 2nd row:", arr[1])
print("Entire 3rd column:", arr[:, 2])
```

```python
# Why NumPy is Better

# Python Loop

a = [1, 2, 3, 4]
b = [10, 20, 30, 40]

result = []

for i in range(len(a)):
    result.append(a[i] * b[i])

print(result)
```

```python
# NumPy Version

a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])

result = a * b

print(result)
```

```python
# Vector Addition

u = np.array([1, 0])
v = np.array([0, 1])

z = u + v

print("u =", u)
print("v =", v)
print("z =", z)
```

```python
# Vector Subtraction

u = np.array([1, 0])
v = np.array([0, 1])

z = u - v

print(z)
```

```python
# Scalar Multiplication

y = np.array([1, 2])

print("Original:", y)

print("Multiply by 2:", 2 * y)
print("Multiply by 5:", 5 * y)
```

```python
# Basic Arithmetic on Arrays

arr = np.array([10, 20, 30, 40])

print("Original:", arr)

print("Add 2      :", arr + 2)
print("Subtract 5 :", arr - 5)
print("Multiply 3 :", arr * 3)
print("Divide by 2:", arr / 2)
```
