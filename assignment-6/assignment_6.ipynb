```python
# Question 1

import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# (a) Add 2°C
updatedTemperature = temperature + 2
print("1(a):", updatedTemperature)

# (b) Celsius to Fahrenheit
fahrenheit = (9 / 5) * temperature + 32
print("1(b):", fahrenheit)

# (c) Readings greater than 32°C
greaterThan32 = temperature[temperature > 32]
print("1(c):", greaterThan32)

# (d) Count readings greater than 32°C
count = np.sum(temperature > 32)
print("1(d):", count)

# (e) Explanation
print("\n1(e):")
print("Vectorization performs operations on the entire array without loops")
print("Boolean indexing filters values directly using conditions")
print("Both methods are faster and more efficient than explicit iteration.")
```

```python
# Question 2

import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# (a)
totalSteps = np.sum(steps)
print("2(a):", totalSteps)

# (b)
meanSteps = np.mean(steps)
print("2(b):", meanSteps)

# (c)
print("2(c) Maximum:", np.max(steps))
print("2(c) Minimum:", np.min(steps))

# (d)
dayWiseTotal = np.sum(steps, axis=0)
print("2(d):", dayWiseTotal)

# (e)
userWiseTotal = np.sum(steps, axis=1)
print("2(e):", userWiseTotal)

# (f)
maxPosition = np.unravel_index(np.argmax(steps), steps.shape)
print("2(f):", maxPosition)
```

```python
# Question 4

import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# (a)
print("4(a) Shape:", X.shape)
print("4(a) Dimensions:", X.ndim)

# (b)
transposeMatrix = X.T
print("\n4(b) Transpose:")
print(transposeMatrix)
print("Transpose converts rows into columns and columns into rows.")

# (c)
xtx = transposeMatrix @ X
print("\n4(c) X.T @ X:")
print(xtx)

# (d)
inverseMatrix = np.linalg.inv(xtx)
print("\n4(d) Inverse:")
print(inverseMatrix)

# (e)
beta = inverseMatrix @ transposeMatrix @ y
print("\n4(e) Coefficients:")
print(beta)

# (f)
print("\n4(f):")
print("Coefficient 1 -> Effect of sleep hours")
print("Coefficient 2 -> Effect of activity level")
print("Coefficient 3 -> Effect of stress level")
print("Each value shows the change in assistance score for one unit increase in that feature.")

# (g)
newUser = np.array([5, 40, 7])
predictedScore = newUser @ beta
print("\n4(g) Predicted Assistance Score:", predictedScore)
```
