```python
import pandas as pd
```

```python
# ============================================================
# 1: IMPORT PANDAS + CREATE EMPLOYEE DATAFRAME
# ============================================================

df = pd.DataFrame({
    "employee": ["Amit", "Neha", "Raj", "Priya", "Amit", "Neha",
                 "Raj", "Priya", "Karan", "Karan"],
    "department": ["IT", "IT", "HR", "HR", "IT", "IT",
                   "HR", "HR", "Sales", "Sales"],
    "salary": [70000, 85000, 60000, 75000, 90000, 80000,
               65000, 70000, 55000, 65000],
    "region": ["North", "North", "South", "South", "North", "North",
               "South", "South", "West", "West"],
    "performance": [82, 91, 76, 88, 95, 86, 80, 90, 72, 85]
})

print(df)
```

```python
# ============================================================
# 2: BASIC INSPECTION
# ============================================================

print("First 5 rows:")
print(df.head())

print("\nDataFrame information:")
df.info()

print("\nColumn names:")
print(df.columns)
```

```python
# ============================================================
# 3: GROUPBY()
# ============================================================

grouped = df.groupby("department")
print(grouped)

print(grouped.groups)

print(grouped.get_group("IT"))

# groupby creates groups; we perform an operation next.
```

```python
# ============================================================
# 4: GROUPBY + COUNT()
# ============================================================

result = df.groupby("department")["salary"].count()
print(result)

# Basically, what we are doing is df -> groupby dept -> select salary -> count
```

```python
# ============================================================
# 5: GROUPBY BY REGION
# ============================================================

result = df.groupby("region")["salary"].count()
print(result)
```

```python
# ============================================================
# 6: GROUPBY PERFORMANCE BY REGION
# ============================================================

result = df.groupby("region")["performance"].count()
print(result)
```

```python
# ============================================================
# 7: GROUPBY + MEAN()
# ============================================================

result = df.groupby("department")["salary"].mean()
print(result)
```

```python
# ============================================================
# 8: GROUPBY + APPLY()
# ============================================================

result = df.groupby("department")["salary"].apply(
    lambda x: x.mean()
)

print(result)
```

```python
# ============================================================
# 9: UNDERSTANDING LAMBDA
# ============================================================

def square(x):
    return x * x

print("Normal function:", square(5))

square_lambda = lambda x: x * x

print("Lambda function:", square_lambda(5))
```

```python
# ============================================================
# 10: TRANSFORM()
# ============================================================

df["department_average_salary"] = (
    df.groupby("department")["salary"]
      .transform("mean")
)

print(df)


# Compare groupby mean with transform
print("\nGroupby mean:")
print(df.groupby("department")["salary"].mean())

print("\nTransform:")
print(df.groupby("department")["salary"].transform("mean"))
```

```python
# ============================================================
# 11: REMOVE DEMONSTRATION COLUMN
# ============================================================

df = df.drop(columns=["department_average_salary"])
print(df)
```

```python
# ============================================================
# 12: NLARGEST()
# ============================================================

result = df.nlargest(3, "salary")
print(result)
```

```python
# ============================================================
# 13: SECOND-HIGHEST SALARY IN EACH DEPARTMENT
# ============================================================

result = df.groupby("department")["salary"].apply(
    lambda x: x.nlargest(2).min()
)

print(result)
```

```python
# ============================================================
# 14: APPLY() VS VECTORIZATION
# ============================================================

demo_df = df.copy()

# Using apply()
demo_df["bonus_apply"] = demo_df["salary"].apply(
    lambda x: x * 0.10
)

# Using vectorization
demo_df["bonus_vectorized"] = demo_df["salary"] * 0.10

print(demo_df[
    ["employee", "salary", "bonus_apply", "bonus_vectorized"]
])
```

```python
# ============================================================
# 15: EXPLODE()
# ============================================================

student_df = pd.DataFrame({
    "student": ["A", "B"],
    "subjects": [
        ["Math", "Sci"],
        ["Eng"]
    ]
})

print("Before explode:")
print(student_df)

print("\nAfter explode:")

exploded_df = student_df.explode("subjects")
print(exploded_df)
```

```python
# ============================================================
# 16: MERGE()
# ============================================================

df1 = pd.DataFrame({
    "user_id": [1, 2, 3],
    "name": ["Alice", "Bob", "Charlie"]
})

df2 = pd.DataFrame({
    "user_id": [2, 3, 4],
    "spend": [50, 120, 10]
})

print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)
```

```python
# ============================================================
# 17: INNER JOIN
# ============================================================

result = pd.merge(
    df1,
    df2,
    on="user_id",
    how="inner"
)

print(result)
```

```python
# ============================================================
# 18: DIFFERENT TYPES OF MERGE
# ============================================================

print("INNER JOIN:")
print(pd.merge(df1, df2, on="user_id", how="inner"))

print("\nLEFT JOIN:")
print(pd.merge(df1, df2, on="user_id", how="left"))

print("\nRIGHT JOIN:")
print(pd.merge(df1, df2, on="user_id", how="right"))

print("\nOUTER JOIN:")
print(pd.merge(df1, df2, on="user_id", how="outer"))
```

```python
# ============================================================
# 19: DEFAULT MERGE
# ============================================================

# If how= is not specified, merge uses inner join by default.

result = pd.merge(df1, df2, on="user_id")
print(result)
```

```python
# ============================================================
# 20: SHIFT()
# ============================================================

status_df = pd.DataFrame({
    "day": [1, 2, 3, 4, 5, 6, 7],
    "status": [
        "Up", "Up", "Up",
        "Down", "Down",
        "Up", "Up"
    ]
})

print(status_df)

status_df["previous"] = status_df["status"].shift()

print("\nAfter shift():")
print(status_df)
```

```python
# ============================================================
# 21: DETECT STATUS CHANGES
# ============================================================

status_change = (
    status_df["status"] != status_df["status"].shift()
)

print(status_change)
```

```python
# ============================================================
# 22: SHIFT() + CUMSUM() FOR STREAK IDs
# ============================================================

status_df["streak_id"] = (
    (status_df["status"] != status_df["status"].shift())
    .cumsum()
)

print(status_df)
```

```python
# ============================================================
# 23: UNDERSTAND THE STREAK LOGIC
# ============================================================

change = status_df["status"] != status_df["status"].shift()

print("Status changed:")
print(change)

print("\nTrue/False as numbers:")
print(change.astype(int))

print("\nCumulative sum:")
print(change.cumsum())
```

```python
# ============================================================
# 24: MAX()
# ============================================================

print("Maximum performance:")
print(df["performance"].max())
```

```python
# ============================================================
# 25: IDXMAX()
# ============================================================

print("Index of maximum performance:")
print(df["performance"].idxmax())
```

```python
# ============================================================
# 26: FIND COMPLETE ROW USING IDXMAX()
# ============================================================

result = df.loc[df["performance"].idxmax()]

print(result)
```

```python
# ============================================================
# 27: MAX() VS IDXMAX() SUMMARY
# ============================================================

print("Maximum value:")
print(df["performance"].max())

print("\nIndex of maximum value:")
print(df["performance"].idxmax())

print("\nComplete row containing maximum:")
print(df.loc[df["performance"].idxmax()])
```
